```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
from sklearn.preprocessing import StandardScaler, MinMaxScaler

train_data = pd.read_csv("cmapss_data/train_FD001.txt", sep="\s+", header=None)
train_data.shape
```

```python
# 設定圖表大小
plt.figure(figsize=(16, 12))

# 遍歷每個欄位（跳過前兩個欄位 0 和 1，因為它們是引擎號和循環數）
for col in train_data.columns[5:]:
    plt.figure(figsize=(12, 4))
    plt.plot(train_data[col], linewidth=0.8)
    plt.title(f'Column {col} - Variation Over Time')
    plt.xlabel('Cycle')
    plt.ylabel('Value')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```
